mrr=1.0 (Mean Reciprocal Rank): Es un resultado perfecto. Significa que, en promedio, la primera respuesta devuelta por el sistema fue la correcta. El MRR evalúa la posición del primer resultado relevante; un 1.0 indica que siempre estuvo en el puesto número 1.

ndcg=0.9823 (Normalized Discounted Cumulative Gain): Un puntaje de ~98.2%. Esta métrica mide la calidad general del ranking, penalizando al sistema si los resultados más relevantes aparecen abajo. Al estar tan cerca de 1.0, significa que el orden de los resultados devueltos es casi ideal.

keywords_found=3 y total_keywords=3: El sistema logró encontrar las 3 palabras clave o conceptos objetivos que se estaban buscando en la evaluación.

keyword_coverage=100.0: Directamente relacionado con lo anterior. El sistema tuvo una cobertura perfecta (100%), lo que significa que no se dejó ningún término clave importante sin recuperar.
