# Crear una pipeline local RAG desde cero

## ¿Qué es RAG?

RAG se ocupa de aumentar la recuperación y generación. 

La meta del RAG is adquirir informacion y enviarla a un LLM que pueda generar una salida vase en esa información.

* Retrieval - Encontrar información relevante dada una query, e.j. "¿Va a llover lo suficiente la próxima semana?" -> recupera los pasajes del texto relacionado con los las lluvias para la próxima semana.
* Augmented - Adquir la informacion relevante y aumentada para nuestra entrada (prompt) para un LLM con eso la información relevante.
* Generation - Adquiera los primeros dos pasos y se lo envía a un LLM para generar salidas

## ¿Por qué utilizar RAG?
La meta principal de RAG es mejorar la generación de la salida LLMs.

1. **Prevenir las alucinaciones**: los LLM son increíbles, pero son propensos a sufrir alucinaciones, es decir, a generar algo que *parece* correcto pero no lo es. Los flujos de trabajo RAG pueden ayudar a los LLM a generar resultados más objetivos al proporcionarles entradas objetivas (recuperadas). E incluso si la respuesta generada por un flujo de trabajo RAG no parece correcta, gracias a la recuperación, también se tiene acceso a las fuentes de las que procede.
2. **Trabajar con datos personalizados**: muchos LLM básicos se entrenan con datos de texto a escala de Internet. Esto significa que tienen una gran capacidad para modelar el lenguaje; sin embargo, a menudo carecen de conocimientos específicos. Los sistemas RAG pueden proporcionar a los LLM datos específicos de un dominio, como información médica o documentación empresarial, y así personalizar sus resultados para adaptarlos a casos de uso específicos.


## ¿Por qué utilizar en local?

Es más divertido que simplemente depender de prestaciones de servicios externos.

Garantiza la privacidad, la velocidad y el coste.

* Privacidad - Al contar con información privada no es necesario enviarla a una API. Es por ello, que quiero que un LLM y ejecutarlo en mi propio hardware.
* Velocidad - En cuanto a la velocidad, significa que no tendrás que esperar necesariamente a que se vacíe la cola de una API o a que se resuelva un tiempo de inactividad; si tu hardware está funcionamiento, el proceso puede ejecutarse.
* Coste - En cuanto a la velocidad, significa que no tendrás que esperar necesariamente a que se vacíe la cola de una API o a que se resuelva un tiempo
inactividad; si tu hardware está en funcionamiento, el proceso puede ejecutarse.

Y desde el punto de vista del cost, ejecutar el proceso en tu propio hardware suele suponer un coste inicial más elevado, pero a partir de ahí los costes
son mínimos o nulos.

En cuanto al rendimiento, es posible que las API de LLM sigan ofreciendo mejores resultados que un modelo de código abierto ejecutando localmente en 
tareas generales, pero cada vez hay más ejemplos de modelos más pequeños y especializados que superan a los modelos más grandes.
